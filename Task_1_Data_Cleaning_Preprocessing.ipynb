```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

missing_table = missing_table[missing_table["Missing Values"] > 0]

missing_table
```

```python
median_income = df["Income"].median()

print("Median Income:", median_income)
```

```python
df["Income"] = df["Income"].fillna(median_income)
print(df["Income"].isnull().sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print(df.columns.tolist())
```

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    dayfirst=True,
    errors="coerce"
)
print(df["dt_customer"].dtype)
```

```python
print("Invalid dates:", df["dt_customer"].isnull().sum())
```

```python
print(df["education"].value_counts())
```

```python
df["education"] = df["education"].str.strip()
```

```python
print(df.dtypes)
```

```python
integer_columns = [
    "id",
    "year_birth",
    "kidhome",
    "teenhome",
    "recency",
    "mnt_wines",
    "mnt_fruits",
    "mnt_meatproducts",
    "mnt_fishproducts",
    "mnt_sweetproducts",
    "mnt_goldprods"
]

for col in integer_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

# Task 1: Data Cleaning and Preprocessing

## Dataset
Customer Personality Analysis – Kaggle

## Objective
To clean and preprocess a raw customer dataset containing missing values,
duplicate records, inconsistent text formatting, date formatting issues,
and data type inconsistencies.

## Tools Used
- Python
- Pandas
- NumPy
- Google Colab

## Data Cleaning Steps Performed

1. Loaded the raw dataset using Pandas.
2. Inspected the dataset using head(), shape, info(), and dtypes.
3. Identified missing values using isnull().
4. Handled missing values in the Income column using the median.
5. Checked and removed duplicate rows.
6. Standardized column names to lowercase with underscores.
7. Removed unnecessary spaces from categorical values.
8. Converted Dt_Customer into datetime format.
9. Checked and corrected numerical data types.
10. Performed final validation for missing values and duplicates.
11. Exported the cleaned dataset as a CSV file.

## Deliverable

The cleaned dataset was exported as:

`cleaned_customer_personality.csv`

## Conclusion

The raw customer dataset was successfully cleaned and preprocessed.
Missing values were handled, duplicate records were removed,
column names were standardized, dates were converted to a consistent
datetime format, and data types were checked. The resulting dataset
is ready for further analysis.
